```python
# Start with knowing about 0 murders
total = 0

# Go through every city, saying "how many murders???"
# then adding that to my total
for murder_count in murders.values():
    total = total + murder_count
    print("Found", murder_count, "new murders, new total is", total)

# Print out the total
print("Total murders:", total)
```
